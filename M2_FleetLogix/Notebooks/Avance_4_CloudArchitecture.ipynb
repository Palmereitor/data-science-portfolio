# Module 4 - Cloud Architecture

**Objective:** design an AWS architecture with RDS, S3, Lambda, DynamoDB, API Gateway, and CloudWatch for the FleetLogix production system.


## Cloud Flow

Mobile app → API Gateway → Lambda → DynamoDB/RDS/S3 → CloudWatch/SNS. Snowflake remains as the analytics layer connected via S3 data lake.

**Architecture principles:**
- Serverless-first: Lambda functions for event processing
- Separation of OLTP (RDS/DynamoDB) and OLAP (Snowflake) workloads
- S3 as data lake for raw events and model artifacts

## Cost Estimation
